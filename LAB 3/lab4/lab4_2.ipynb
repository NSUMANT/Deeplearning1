# Object Detection using VGG16 & Selective Search
This notebook demonstrates an object detection pipeline using region proposals from Selective Search (or a sliding window fallback) and classification with a pre-trained VGG16 model.

```python
# ---------------------------------------------------------------------------
# 0. INSTALL OPENCV CONTRIB (RUN ONLY IF REQUIRED)
# ---------------------------------------------------------------------------
import os
import subprocess

try:
    import cv2
    ss_ready = (
        hasattr(cv2, "ximgproc")
        and hasattr(cv2.ximgproc, "segmentation")
    )
except Exception:
    ss_ready = False

if not ss_ready:
    subprocess.run(
        "pip uninstall -y -q opencv-python opencv-python-headless "
        "opencv-contrib-python opencv-contrib-python-headless",
        shell=True
    )
    subprocess.run(
        "pip install -q opencv-contrib-python",
        shell=True
    )
    os.kill(os.getpid(), 9)  # Runtime restarts; run the cell again
```

```python
# ---------------------------------------------------------------------------
# 1. IMPORTS
# ---------------------------------------------------------------------------
import tensorflow as tf
from tensorflow.keras.applications import VGG16
from tensorflow.keras.applications.vgg16 import (
    preprocess_input,
    decode_predictions
)
import numpy as np
import cv2
from google.colab import files
from google.colab.patches import cv2_imshow
```

```python
# ---------------------------------------------------------------------------
# 2. LOAD PRE-TRAINED VGG16 CLASSIFIER
# ---------------------------------------------------------------------------
model = VGG16(weights="imagenet", include_top=True)
print("VGG16 model loaded successfully!")
```

```python
# ---------------------------------------------------------------------------
# 3. REGION PROPOSALS USING SELECTIVE SEARCH
# ---------------------------------------------------------------------------
def get_region_proposals(img, max_regions=200):
    try:
        ss = cv2.ximgproc.segmentation.createSelectiveSearchSegmentation()
        ss.setBaseImage(img)
        ss.switchToSelectiveSearchFast()

        boxes = [
            (x, y, w, h)
            for (x, y, w, h) in ss.process()
            if w > 40 and h > 40
        ]

        if boxes:
            return boxes[:max_regions]

    except Exception:
        pass

    # Fallback: multi-scale sliding windows
    H, W = img.shape[:2]
    boxes = []

    for s in (0.6, 0.4, 0.25):
        bw, bh = int(W * s), int(H * s)

        if bw <= 0 or bh <= 0:
            continue

        for y in range(0, H - bh + 1, max(1, bh // 2)):
            for x in range(0, W - bw + 1, max(1, bw // 2)):
                boxes.append((x, y, bw, bh))

    return boxes[:max_regions]
```

```python
# ---------------------------------------------------------------------------
# 4. NON-MAXIMUM SUPPRESSION
# ---------------------------------------------------------------------------
def non_max_suppression(dets, iou_thr=0.3):
    if not dets:
        return []

    boxes = np.array(
        [[x, y, x + w, y + h] for (x, y, w, h), _, _ in dets],
        dtype=float
    )

    scores = np.array([score for _, _, score in dets])
    order = scores.argsort()[::-1]
    keep = []

    while len(order):
        i = order[0]
        keep.append(i)
        rest = order[1:]

        if len(rest) == 0:
            break

        xx1 = np.maximum(boxes[i, 0], boxes[rest, 0])
        yy1 = np.maximum(boxes[i, 1], boxes[rest, 1])
        xx2 = np.minimum(boxes[i, 2], boxes[rest, 2])
        yy2 = np.minimum(boxes[i, 3], boxes[rest, 3])

        w = np.clip(xx2 - xx1, 0, None)
        h = np.clip(yy2 - yy1, 0, None)
        inter = w * h

        area_i = (
            (boxes[i, 2] - boxes[i, 0])
            * (boxes[i, 3] - boxes[i, 1])
        )

        area_r = (
            (boxes[rest, 2] - boxes[rest, 0])
            * (boxes[rest, 3] - boxes[rest, 1])
        )

        iou = inter / (area_i + area_r - inter + 1e-9)
        order = rest[iou < iou_thr]

    return [dets[i] for i in keep]
```

```python
# ---------------------------------------------------------------------------
# 5. UPLOAD IMAGE
# ---------------------------------------------------------------------------
uploaded = files.upload()
image_path = next(iter(uploaded))

img = cv2.imread(image_path)

if img is None:
    raise ValueError("Could not read the uploaded image.")

H0, W0 = img.shape[:2]

# Resize image for faster processing
img = cv2.resize(
    img,
    (600, max(1, int(H0 * 600.0 / W0)))
)
```

```python
# ---------------------------------------------------------------------------
# 6. GENERATE REGION PROPOSALS
# ---------------------------------------------------------------------------
boxes = get_region_proposals(img, max_regions=200)

print("Number of region proposals:", len(boxes))

if not boxes:
    raise ValueError("No region proposals were generated.")
```

```python
# ---------------------------------------------------------------------------
# 7. CLASSIFY REGIONS USING VGG16
# ---------------------------------------------------------------------------
crops = np.array(
    [
        cv2.resize(img[y:y+h, x:x+w], (224, 224))
        for (x, y, w, h) in boxes
    ],
    dtype=np.float32
)

# Predict all crops in batches
preds = model.predict(
    preprocess_input(crops),
    batch_size=32,
    verbose=0
)
```

```python
# ---------------------------------------------------------------------------
# 8. FILTER DETECTIONS BY CONFIDENCE & 9. REMOVE DUPLICATE BOXES
# ---------------------------------------------------------------------------
detections = []

for (x, y, w, h), p in zip(boxes, preds):
    _, label, prob = decode_predictions(p[None, :], top=1)[0][0]

    if prob > 0.90:
        detections.append(((x, y, w, h), label, float(prob)))

# Non-maximum suppression to filter overlapping windows
detections = non_max_suppression(detections, iou_thr=0.3)

print("Objects detected:", len(detections))
```

```python
# ---------------------------------------------------------------------------
# 10. DRAW BOUNDING BOXES AND DISPLAY RESULT
# ---------------------------------------------------------------------------
img_output = img.copy()
for (x, y, w, h), label, score in detections:
    cv2.rectangle(
        img_output,
        (x, y),
        (x + w, y + h),
        (0, 255, 0),
        2
    )

    cv2.putText(
        img_output,
        f"{label}: {score:.2f}",
        (x, max(18, y - 6)),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.6,
        (0, 255, 0),
        2
    )

cv2_imshow(img_output)
```
